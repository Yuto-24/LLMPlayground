```python
import asyncio
import time
from datetime import datetime as dt


async def async_func(num):
    print(f"func{num}: [{dt.now()}] start async_func")
    await asyncio.sleep(2)
    print(f"func{num}: [{dt.now()}] end async_func")


async def main_async(num):
    print(f"main{num}: [{dt.now()}] start main")
    await async_func(num)
    await async_func(num+1)
    print(f"main{num}: [{dt.now()}] end main")
```

```python
loop = asyncio.get_event_loop()
loop.create_task(main_async(10))
loop.create_task(main_async(20))
```
